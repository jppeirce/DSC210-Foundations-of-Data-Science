### 1.2 Numbers alone can be deceiving

#### **Class Example 5.1 - Anscombe's Quartet**

Load *Anscombe's Quartet* (Anscombe, 1973): four small datasets, where each produces the same summary statistics (mean, standard deviation, and correlation).

````python
# RUN-TOGETHER
import numpy as np
anscombe = sns.load_dataset('anscombe')

# Compute mean of x, mean of y, and correlation for each of the four sets.
for name, group in anscombe.groupby('dataset'):
    mx = group['x'].mean()
    my = group['y'].mean()
    r = np.corrcoef(group['x'], group['y'])[0, 1]
    print(f'Set {name}:  mean x = {mx:.2f}, mean y = {my:.2f}, r = {r:.2f}')
````

> **Predict:** Every one of the four datasets has essentially the **same** mean x, mean y, and correlation. So, predict: do the four datasets *look* the same? Sketch what you expect on paper, then we plot them.

````python
# RUN-TOGETHER
sns.relplot(data=anscombe, x='x', y='y', col='dataset',
            col_wrap=2, height=3)
plt.show()
````

Identical statistics, four completely different stories: one clean line, one curve, one line dragged by a single outlier, one vertical stack.

**Note:** 
- a statistical summary compresses information
- a plot helps reveal details. 

*Both* are necessary for data science. Never trust a data description without looking.

### 1.3 One variable: histograms

**Definition.** A *histogram* is a graphical representation used to visualize the frequency distribution of a dataset. It shows how many data points fall into specific ranges of values.

- **Bins:** The entire range of data is divided into a series of non-overlapping, consecutive intervals called bins (or buckets).
- **Frequency:** The height of the bar for each bin represents the number of data points (frequency) that fall within that interval.
- **Visualization:** Unlike a bar chart (which compares discrete categories), a histogram is used for continuous or quantitative data to show the "shape" of the data.

We can do this in one line in seaborn:

````python
# RUN-TOGETHER
penguins = sns.load_dataset('penguins')
penguins.head()
````

````python
# RUN-TOGETHER
sns.histplot(data=penguins, x='flipper_length_mm')
plt.show()
````

### 1.4 Comparing groups: boxplots

**Definition.** A *boxplot* (or box-and-whisker plot) is a standardized way of displaying the distribution of a quantitative variable based on a five-number summary. It is used to visually identify the spread, skewness, and potential outliers in a numerical dataset. Boxplots are a common tool for comparing distributions across the groups of a categorical variable.

A boxplot consists of several key components derived from the data's distribution:

- Median (Q2): A line inside the box representing the middle value of the dataset (50th percentile).
- First Quartile (Q1): The bottom edge of the box, representing the 25th percentile (the median of the lower half of the data).
- Third Quartile (Q3): The top edge of the box, representing the 75th percentile (the median of the upper half of the data).
- Interquartile Range (IQR): The height of the box itself (Q3 - Q1). It represents the middle 50% of the data.
- Whiskers: Lines extending from the top and bottom of the box. They typically extend to the furthest data points that fall within 1.5 x IQR from the edge of the box.
- Outliers: Individual points plotted beyond the whiskers. These are considered extreme values that fall significantly outside the rest of the data.

````python
# RUN-TOGETHER
sns.boxplot(data=penguins, x='flipper_length_mm', y ='species')
plt.show()
````

### 2.1 Two variables: scatterplots

**Definition.** A *scatterplot* is a diagram that uses dots to represent the values of two different numeric variables. The position of each dot is determined by its x-coordinate (first variable) and y-coordinate (second variable).

The main goal of a scatterplot is to visualize the relationship between two continuous variables, to see whether they change together.

Scatterplots are commonly summarized by:
- Correlation
    - Positive: as x increases, y also increases (the dots trend upward from left to right).
    - Negative: as x increases, y decreases (the dots trend downward).
    - None: the dots appear randomly scattered with no discernible trend.
- Strength of relationship:
    - Strong: the dots are tightly clustered around a clear line or curve.
    - Weak: the dots are widely dispersed, making the trend harder to see.
- Form (shape):
    - Linear: the relationship follows a straight line.
    - Non-linear: the relationship follows a curve.
- Outliers:
    - Points that fall far from the general cluster, indicating unusual observations or errors in data collection.

````python
# RUN-TOGETHER
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g')
plt.show()
````

> **Predict, then run.** We know these two variables are strongly correlated overall (we saw that in Module 4). If we color by species, will we see one big cloud, or separate clusters?

````python
# RUN-TOGETHER
# Color the points by species to reveal structure.
sns.scatterplot(data=penguins, x='flipper_length_mm',
                y='body_mass_g', hue='species')
plt.title('Body Mass vs Flipper Length by Species')
plt.show()
````

> **Discuss.** The overall trend is there, but each species also forms its own cluster. You have seen what this structure can do: in Module 4, pooling the species *reversed* the bill length versus bill depth correlation, and we named that reversal **Simpson's paradox**. This plot is what that mechanism looks like. If the clusters were arranged differently, the pooled trend could point opposite to every within-species trend, and nothing in a single correlation number would warn you.

````python
# RUN-TOGETHER
# Color the points by species to reveal structure.
sns.scatterplot(data=penguins, x='bill_length_mm',
                y='bill_depth_mm', hue='species')  
plt.title('Bill Length vs Bill Depth by Species')
plt.show()
````

### 2.2 See it all at once: pairplot

A "quick overview" tool for Exploratory Data Analysis (EDA) is the pairplot. `sns.pairplot` creates a grid of plots that visualizes every pairwise relationship between the numeric variables in a dataset simultaneously.

If your dataset has N numeric columns, pairplot creates an NxN grid with two kinds of entries:

1. **The diagonal (univariate plots).** The plots from top-left to bottom-right each show one single variable: its distribution (shape), usually as a histogram or a KDE (kernel density estimate, a smooth curve).
2. **The off-diagonal (bivariate plots).** Every other plot shows the relationship between two different variables, by default as a scatterplot. These are where you spot correlations, clusters, and outliers.

When you meet a new dataset, `pairplot` is the fastest way to get oriented.

````python
# RUN-TOGETHER
sns.pairplot(penguins)
plt.show()
````

#### The power of the hue parameter

The most important feature of pairplot is the `hue` argument. If you provide a categorical column to `hue`, seaborn will:

- color-code the dots in the scatterplots by category, and
- overlay one distribution curve per category in the diagonal plots.

This lets us see whether different groups behave differently across every dimension of the data at once.

````python
# RUN-TOGETHER
sns.pairplot(penguins, hue='species')
plt.show()
````

> **Predict.** On the linear axis, the poorer countries are all squashed against the left edge, hard to distinguish. A **log scale** on the x-axis spreads them out. Predict how the shape changes, then run.

````python
# RUN-TOGETHER
sns.scatterplot(data=gm, x='gdp_cap', y='life_exp')
# switch the x-axis to a logarithmic scale.
plt.xscale('log')
plt.title('Life Expectancy vs GDP per Capita (log x-axis)')
plt.xlabel('GDP per capita (USD, log scale)')
plt.ylabel('Life expectancy (years)')
plt.show()
````

On the log axis the relationship becomes clear and roughly linear: each *multiplicative* step in income buys a similar gain in life expectancy, up to a point.

*The scale you choose is part of the story you tell*, which is exactly why Section 2.4 is about honesty.

## Adding to the Toolkit
---

Every module from here to the final adds rows to one running table. By the last week you will have the whole collection, and the review will be about **choosing** from it: given the feature types you have and the question you were asked, which tool is right?

The columns never change:

| Tool | Question it answers | Feature types it needs | Label needed? | Key assumption | How it fails |
| --- | --- | --- | --- | --- | --- |
| **Histogram** | What shape does one numeric variable have? | one interval or ratio | no | binning does not distort the shape | the bin count changes the story; try two or three |
| **Boxplot** | How do groups compare on a numeric variable? | one numeric, split by one nominal or ordinal | no | the five-number summary is enough | hides multimodality entirely (Activity 5.1) |
| **Bar chart / `countplot`** | How frequent is each category? | one nominal or ordinal | no | categories are countable | says nothing about a numeric variable |
| **Scatterplot** | Are two numeric variables related? | two interval or ratio | no | points are visible, not overplotted | hides group structure unless you add `hue` |
| **Pairplot** | I know nothing about this dataset; orient me. | several numeric, plus a categorical for `hue` | no | the grid stays readable | unreadable beyond roughly six columns |
| **Log scale** | The data spans orders of magnitude. | positive interval or ratio | no | multiplicative steps are the meaningful ones | undefined at zero or below |

**Note:** The first three rows take **one** variable, the next two take **two or more**, and the last is not a plot at all but a change of ruler. When you are stuck choosing, counting your variables narrows the list faster than anything else.
