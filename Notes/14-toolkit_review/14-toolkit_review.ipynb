## 1. The Assembled Toolkit
---

### 1.1 First, the vocabulary that decides everything

Module 2 established the rule the whole table obeys: **a tool is ruled in or out by the measurement scale of the feature, not by how the values happen to be stored.**

| Scale | What it is | Arithmetic allowed? | Example |
| --- | --- | --- | --- |
| **Nominal** | unordered categories | none; counts only | species, payment type, zip code |
| **Ordinal** | ordered categories, unequal gaps | order only; median, percentiles | survey rating, triage level, letter grade |
| **Interval** | equal gaps, no true zero | differences, mean | year, temperature in Celsius |
| **Ratio** | equal gaps, true zero | everything, including ratios | mass, distance, income, count |

Two habits to carry into every scenario below:

1. **Numbers are not automatically numerical.** Zip codes, jersey numbers, and `pclass` are labels wearing digits.
2. **Ask what a mean would mean.** If the answer is "nothing", the feature is nominal and most numeric tools are out.

### 1.2 Describing one variable

| Tool | Question it answers | Feature types it needs | Label needed? | Key assumption | How it fails |
| --- | --- | --- | --- | --- | --- |
| **Mean** | What is a typical value? | interval or ratio | no | values are quantities | meaningless on nominal; dragged by skew |
| **Median** | What is a typical value? | ordinal, interval, ratio | no | values can be ordered | ignores magnitude |
| **Mode / `value_counts`** | Which category is most common? | any; the only choice for nominal | no | categories are discrete | uninformative on continuous data |
| **Histogram** | What shape does this variable have? | interval or ratio | no | binning does not distort | bin count changes the story |
| **Boxplot** | Where are the centre, spread, and extremes? | interval or ratio | no | five numbers are enough | hides multimodality completely |
| **1.5 x IQR fence** | Which values are unusually far out? | interval or ratio | no | roughly unimodal | flags legitimate subpopulations |

### 1.3 Relating two or more variables

| Tool | Question it answers | Feature types it needs | Label needed? | Key assumption | How it fails |
| --- | --- | --- | --- | --- | --- |
| **Grouped boxplot** | How do groups compare on a numeric variable? | numeric, split by categorical | no | groups are meaningful | small groups are jumpy |
| **`groupby` summary** | Does a numeric variable differ across groups? | numeric, split by categorical | no | groups large enough | check the counts |
| **Crosstab** | Are two categorical variables associated? | two nominal or ordinal | no | cells have enough counts | raw counts mislead; use `normalize` |
| **Scatterplot** | Are two numeric variables related? | two interval or ratio | no | not overplotted | hides groups without `hue` |
| **Correlation $r$** | How strongly, in a straight line? | two interval or ratio | no | relationship is **linear** | blind to curves; outlier-sensitive; not causation |
| **Pairplot** | Orient me in a new dataset. | several numeric plus a categorical | no | grid stays readable | unusable beyond ~6 columns |

### 1.4 Finding structure without labels

| Tool | Question it answers | Feature types it needs | Label needed? | Key assumption | How it fails |
| --- | --- | --- | --- | --- | --- |
| **k-means** | Are there groups, if I commit to `k`? | interval/ratio, **scaled** | **no** | round, similar-sized clusters | needs `k`; init-sensitive; fails on crescents |
| **Hierarchical** | What is the nested structure at every scale? | anything with a distance, **scaled** | **no** | linkage matches "close" | linkage changes the tree; scales poorly |
| **DBSCAN** | Dense groups, and which points belong to none? | interval/ratio, **scaled** | **no** | clusters are dense regions | `eps` hard to choose |
| **PCA** | Can few columns stand in for many? | several interval/ratio, **scaled** | **no** | variance is what matters; linear | components need interpreting |

### 1.5 Predicting a label

| Tool | Question it answers | Feature types it needs | Label needed? | Key assumption | How it fails |
| --- | --- | --- | --- | --- | --- |
| **Train/test split** | Will this work on unseen data? | any | **yes** | test set stays untouched | leakage makes estimates optimistic |
| **k-NN classifier** | Which category is this new case? | interval/ratio, **scaled** | **yes**, categorical | nearby cases share labels | `k`; scaling; slow to predict |
| **Decision tree** | Which category, by what readable rule? | any; **no scaling** | **yes**, categorical | axis-aligned cuts suffice | overfits without depth limit |
| **Confusion matrix, precision, recall** | What kind of mistakes? | predictions plus true labels | **yes** | classes are meaningful | accuracy hides everything when imbalanced |

## 4. What the Toolkit Does Not Contain
---

Worth saying plainly before the final.

**Causation.** Not one tool in Section 1 establishes that A causes B. Correlation, group differences, and classification accuracy are all compatible with a lurking variable. Module 4's Simpson's paradox and Module 7's tolls-and-distance discussion were both about this.

**Whether the data can answer the question.** Module 2's "right data" idea and Module 7's cash-tip artifact both showed accurate data that could not answer the question asked. No tool detects that; a person has to.

**Whether the sample represents anything.** Homework 3's exoplanet catalogue described the detection process at least as much as the galaxy, and Module 6's most-streamed songs list was a survivor list. Selection effects survive every tool in the table.

**What the result means.** A cluster is not a customer segment until a person who knows the business says so. A principal component is unnamed until someone reads the loadings. This is the Substantive Expertise circle from Module 1, and it is the one part of the job the toolkit cannot do for you.
