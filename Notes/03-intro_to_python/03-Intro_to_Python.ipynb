### 1.2 Variable assignment

### Class Example 3.1 - Savings over time

> **Predict first.** You start with \$100 and save \$10 per month. How much after 4 months? Predict, then we compute it together.

````python
# RUN-TOGETHER
# Suppose you start with $100 in savings.
# How much money would you have saved four months from now,
# if you saved $10 each month?

initial_savings = 100
monthly_savings = 10
months = 4

total_savings = initial_savings + monthly_savings * months
print(total_savings)

print(100 + 10 * 4)
````

### 2.2 Indexing and slicing

We reach into a list by position, or index. Python counts from **0**, not 1.

- `x[0]` is the first element, `x[1]` the second.
- `x[-1]` is the last element (negative counts from the end).
- `x[start:stop]` is a **slice**: from `start`, up to but **not including** `stop`.

````python
# RUN-TOGETHER
bill_lengths = [39.1, 39.5, 40.3, 36.7, 39.3, 38.9]
print(bill_lengths[0])    # first  -> 39.1
print(bill_lengths[-1])   # last   -> 38.9
print(bill_lengths[0:3])  # first three -> [39.1, 39.5, 40.3]
print(bill_lengths[:3])   # same thing: omitting start means 'from the beginning'
````

### 2.5 The copy trap (aliasing)

In data work, accidentally altering your dataset is hard to notice. A common silent bug for new Python users is creating an *alias* instead of a *copy*.

Writing `b = a` for a list does **not** make a copy. It makes a second name for the *same* list.

````python
# RUN-TOGETHER
original = [1, 2, 3, 4]
alias = original          # NOT a copy: same list, two names
alias.append(99)

print('alias   :', alias)
print('original:', original)

# Keep this in mind especially when dealing with raw (original) data.
````

````python
# RUN-TOGETHER
# The fix: make a real copy with .copy() or a full slice [:]
original = [1, 2, 3, 4]
real_copy = original.copy()    # independent copy
real_copy.append(99)

print('real_copy:', real_copy)
print('original :', original)
````

### Class Example 3.4 - Dot notation

Suppose we want to capitalize `species`. Does calling it like a plain function work?

````python
# RUN-TOGETHER
# This cell is SUPPOSED to fail. Read the error message.
species = 'gentoo'
capitalize(species)
````

````python
# RUN-TOGETHER
# Instead, to call the method capitalize(), we use "dot notation":
print(species.capitalize())   # call method capitalize() on species

# The method capitalize() belongs to all string objects.
# It is automatically available as soon as the object is created.
print(species.upper())

# Some methods take inputs. For example, with a list object:
nums = [3, 1, 4, 1, 5]
print(nums.count(1))   # list method: how many 1s?
print(nums.index(1))   # position of the first 1
nums.append(42)
print(nums)
````

````python
# RUN-TOGETHER
# Depending on the object type, a method of the same name behaves differently.
species = "gentoo"
print(species.index("t"))
print(species.count("o"))

# A string behaves in some ways like a sequence of characters:
# species -> ['g', 'e', 'n', 't', 'o', 'o']
````

### 3.1 What is a package?

What about other methods? Something a little more niche, nuanced, or specialized?

Not everybody needs every function that has ever been written, so most are not included in a Python distribution. This is where *packages* come in.

The data science stack

```
ASK  ->  GET  ->  EXPLORE  ->  MODEL  ->  COMMUNICATE
 ^                                            |
 |________ findings raise new questions ______|
```

has a few you will use constantly:

| Package | Gives us | Used in |
| --- | --- | --- |
| `numpy` | fast numeric arrays | storing data (GET) |
| `matplotlib` / `seaborn` | data visualization | (EXPLORE) |
| `scikit-learn` | machine learning tools | (MODEL) |

### 3.2 Installing packages

Depending on how you access Python (cloud versions, local versions), you may need to install a package before you can import it. See [the official installation guide](https://packaging.python.org/en/latest/tutorials/installing-packages/) if needed.

In Jupyter notebooks accessed through Google Colab, many common packages are already available and nothing needs to be installed.

### 3.3 Importing packages

Before we can use functions from a package, we need to import the package (or a specific piece of it) with an `import` statement.

````python
# RUN-TOGETHER
# This cell is SUPPOSED to fail. Read the error message.
import numpy
array([1, 2, 3])

# array() is a function IN the numpy package.
# We have to say where it lives.
````

````python
# RUN-TOGETHER
numpy.array([1, 2, 3])
# Try print() with this. Are the outputs different?
print(numpy.array([1, 2, 3]))
````

````python
# RUN-TOGETHER
# Less typing: give the package a short nickname
import numpy as np
print(np.array([1, 2, 3]))
````

````python
# RUN-TOGETHER
# Be more selective: import one function directly.
# Then you do NOT use the numpy. or np. prefix.
from numpy import array
print(array([1, 2, 3]))
````

### 3.4 Which option is best?

- `import numpy` followed by `numpy.array()` is the clearest version in code, but requires more typing
- `import numpy as np` followed by `np.array()` requires some human memory
- `from numpy import array` followed by `array()` requires the least typing, but hides some detail (and requires more human memory)

The structure `import numpy as np` is used very frequently. There are common shorthand names for many of the packages we will see.
